```python
#the __init__() function
class Person:
  def __init__(self,name,age):
    self.name = name
    self.age = age

p1 = Person("John",36)
print(p1.name)
print(p1.age)
```

```python
#the __str__() function
class Person:
  def __init__(self,name,age):
    self.name = name
    self.age = age
  def __str__(self):
    return f"{self.name}({self.age})"

p1= Person("John",36)
print(p1)
```

```python
#obj methods
class Person:
  def __init__(self,name,age):
    self.name = name
    self.age = age
  def myfunc(self):
    print("Hello my name is " + self.name)

p1 = Person("Shahed",29)

p1.myfunc()
```

```python
#encapsulation
class BankAccount:
  def __init__(self,balance):
    self.__balance = balance
  def get_balance(self):
    return self.__balance
  def set_balance(self,amount):
    if amount < 0:
      print("Invalid amount")
    else:
      self.__balance = amount

acc= BankAccount(1000)
acc.set_balance(-500)
print(acc.get_balance())
```
