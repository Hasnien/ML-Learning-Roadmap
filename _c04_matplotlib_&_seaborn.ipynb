```python
import matplotlib.pyplot as plt

days = [1,2,3,4,5]
sales = [11000,12400,8700,14987,13200]
plt.plot(days,sales,marker="o",color="r",linestyle="--")
plt.grid()
plt.xlabel("days")
plt.ylabel("sales")
plt.title("Daily Sales")
plt.show()
```

```python
import matplotlib.pyplot as plt
import numpy as np
import pandas as pd

days=[1,2,3,4,5,6,7]
sales_2025 = [123231,123221,143212,123143,142223,121232,120009]
sales_2026 = [129878,137898,111299,100987,145678,135467,111445]
plt.plot(days,sales_2025,marker="o",label="2025")
plt.plot(days,sales_2026,marker="o",label="2026")
plt.grid()
plt.xlabel("days")
plt.ylabel("sales")
plt.title("Sales Comparison")
plt.legend()
plt.show()
```

**Scatter Plot**

```python
#subplots
import matplotlib.pyplot as plt
days = [1,2,3,4,5,6]
sales = [1200,1300,1400,1600,1500,1350]
dept = ["HR","IT","Sales","MIS"]
ages = [21,21,21,22,22,23,24,26,27,28,28,28,28,28,31,31,34,36,39,40]
employees = [10,17,21,43]

plt.subplot(1,3,1)
plt.plot(days,sales)
plt.title("Line")

plt.subplot(1,3,2)
plt.bar(dept,employees)
plt.title("Bar")

plt.subplot(1,3,3)
plt.hist(ages,bins=10)
plt.title("Histogram")

plt.show()
```

**Seaborn**

```python
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt

df = pd.DataFrame({
    "age": [21,21,21,22,22,23,24,26,27,28],
    "salary": [21300,21000,23100,24500,28900,23800,23300,43000,98000,21000],
    "department": ["HR","HR","IT","IT","HR","Admin","Marketing","Sales","CEO","IT"]
})
```

```python
#sns histplot
sns.histplot(data = df, x = "salary" ,kde = True)
plt.title("Salary Distribution")
plt.show()
```

```python
#sns scatterplot
sns.scatterplot(data = df, x = "age" , y = "salary")
plt.title("Age vs Salary")
plt.show()
```

```python
#adding a category with hue
sns.scatterplot(data = df, x = "age" , y = "salary", hue = "department")
plt.title("Age vs Salary")
plt.show()
```

```python
#counterplot
sns.countplot(data = df, x = "department")
plt.title("Department Count")
plt.show()
```

```python
#boxplot
sns.boxplot(data = df, x = "department" , y = "salary")
plt.title("Department vs Salary")
plt.show()
```

```python
#Barplot
sns.barplot(data = df, x = "department" , y = "salary")
plt.title("Department vs Salary")
plt.show()
```
