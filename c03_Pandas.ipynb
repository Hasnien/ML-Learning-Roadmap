**Value Counts**

```python
data = {
    "Name": ["Rahim", "Karim", "Sara", "John", "Aisha", "David"],
    "Age": [25, 32, 28, 22, 35, 29],
    "Salary": [30000, 55000, 45000, 25000, 70000, 48000],
    "Experience": [1, 5, 3, 0, 8, 4],
    "Purchased": ["No", "Yes", "Yes", "No", "Yes", "Yes"]
}

df = pd.DataFrame(data)
df
```

```python
print(df["Purchased"].value_counts())
```

```python
data = {
    "Department": ["IT", "IT", "HR", "HR", "Sales"],
    "Salary": [50000, 60000, 40000, 45000, 55000]
}

df = pd.DataFrame(data)
```

```python
result = df.groupby("Department")["Salary"].mean()
result
```

**Mini project of pandas**

```python
import pandas as pd

data = {
    "Name": ["Rahim", "Karim", "Sara", "John", "Aisha", "David"],
    "Age": [25, 32, 28, 22, 35, 29],
    "Salary": [30000, 55000, 45000, 25000, 70000, 48000],
    "Experience": [1, 5, 3, 0, 8, 4],
    "Purchased": ["No", "Yes", "Yes", "No", "Yes", "Yes"]
}

df = pd.DataFrame(data)

print(df)
```

```python
print(df.shape)
df.info()
```

```python
#Display only:Name & Salary
print(df[["Name","Salary"]])
```

```python
#Find people with salary > 40000.
print(df["Salary"]>40000)

#Find people with experience >= 4.
print(df["Experience"]>=4)
```

```python
#annualsalary
df["annualSalary"] = df["Salary"]*12
print(df["annualSalary"])
```

```python
#average salary,higest salary
print(df["Salary"].mean())
print(df["Salary"].max())
```

```python
print(df["Purchased"].value_counts())
```

```python
print(df.groupby("Purchased")["Salary"].mean())
```
