**Mini Project**

```python
#creating a sample dataset

import numpy as np
scores = np.array([
    [70,80,90],
    [60,75,85],
    [90,95,100],
    [55,65,70],
    [85,90,95]
])
print(scores)
print(scores.ndim)
print(scores.shape)
```

```python
"""Column 0 → Math
Column 1 → English
Column 2 → Science"""

"""Row 0 → Student 1
Row 1 → Student 2
Row 2 → Student 3
Row 3 → Student 4
Row 4 → Student 5"""

#overall average of students
print("Overall Average: ",np.mean(scores))

#highest score
print("Highest Score: ",np.max(scores))
#lowest score
print("Lowest Score: ",np.min(scores))
#median
print("Median Score: ",np.median(scores))
#Standard deviation
print("Standard Deviation: ",np.std(scores))
```

```python
#Subject-wise averages

#math average
print("Math Average: ",np.mean(scores[:,0]))
#eng average
print("English Average: ",np.mean(scores[:,1]))
#science average
print("Science Average: ",np.mean(scores[:,2]))
```

```python
#student wise averag
print("Student 1 Average: ",np.mean(scores[0,:]))
print("Student 2 Average: ",np.mean(scores[1,:]))
print("Student 3 Average: ",np.mean(scores[2,:]))
print("Student 4 Average: ",np.mean(scores[3,:]))
print("Student 5 Average: ",np.mean(scores[4,:]))
```

```python
#find top student
print("Top Student: ",np.max(np.mean(scores,axis=1)))
```

Let's say:

A student passes if their average score is 65 or higher.

```python
passed = np.where(np.mean(scores,axis=1)>=65)
print("Passed: ",passed)
failed = np.where(np.mean(scores,axis=1)<65)
print("Failed: ",failed)
```

Add a 5-point bonus

```python
scores_bonus = scores + 5
print(scores_bonus)
```
